### LLM Benchmarks and Leaderboards
#### How to Compare LLMs
- Open Source or Closed Source:
    - Open Source (e.g. LLaMA) gives you flexibility, full control, and lower cost if self-hosted.
    - Closed Source (e.g. GPT-4) usually offers better performance and ease of use via APIs.
- Context Length:
    - How much input/output the model can handle in one go.
    - For long documents or multi-turn conversation, loger context is better (e.g. Gemini 2.5 Flash has 1M Window)
- Release Data and Knowledge Cut-off 
    - Newer models are trained on more recent data (useful for current events) (e.g.: GPT-3.5's cut-off is early 2023)
- Parameters:
    - More parameters usually = more capacity (e.g. GPT-3 has 175B)
    - But bigger isn't always better for performance or cost-efficiency matters too.
- Training Tokens:
    - Refers to how much text the model was trained on.
    More tokens = broader language understanding and generalization.

### Start-Up Ideas Generator Prompt designing

We need a consistent prompt to give each AI model. This prompt should clearly state:
1. The `context` or personality we want the AI to take (e.g. You are an expert web developer)
2. The `instruction`: generate HTML code for a landing page.
3. The `output indicator` required: spceifically, the full HTML structure for an `index.html` file.  

Lets define our startup idea and the prompt.

````python
# Define the startup name and concpet
startup_name = "Genie_Firma"
startup_concept = "An intelligent CRM system that uses AI to analyze customer interaction \
    , predict needs, and automate presonalized follow-ups. Focus on improving customer \
        retention and sales efficiency for business of  all sizes."
        
````

````python
# Define the core prompt for the LLMs.
# We explicitly ask for HTML code and specify the file name 'index.html'
html_prompt = f"""
You are a helpful AI Assistant acting as a front-end web developer.
Your task is to generate the complete HTML code for a simple, clean, and professional-looking landing page (index.html) for a new startup.

Startup Name : {startup_name}
Concept : {startup_concept}

Please generate ONLY the full HTML code, starting with <!DOCTYPE html> and ending with </html>.
Create a modern, visually appealing landing page with the following:

-- Don't include images in the code. Raw html code with inline css for styling.

1. A sleek header with the startup name in a bold, modern font and a compelling tagline
2. A hero section with a clear value proposition and call-to-action button
3. A features section highlighting 3-4 key benefits with icons or simple visuals
4. A "How it Works" section with numbered steps
5. A testimonials section with fictional customer quotes
6. A pricing section with at least two tiers
7. A professional footer with navigation links and social media icons

Use inline CSS for styling with a modern color palette (primary, secondary, and accent colors). 
Include responsive design elements, subtle animations, and whitespace for readability.
Emphasize AI capabilities, ease of use, and business benefits throughout the copy.
Focus on conversion-optimized marketing messages that highlight pain points and solutions.

Do not include any explanations before or after the code block. Just provide the raw HTML code.
"""
print_markdown("**Core Prompt defined for the LLms:**")
print_markdown(f"> {html_prompt}")
````

### Generate HTML Landing Pages

Start with Locally hosted Model using `Ollama` - `Gemma4:e4b`.
and  Then `OpenAI's` - model `gpt-4o-mini`.
We Will:
1. Call the `chat.completions.create` method.
2. Use our `html_prompt` as the user message.
3. Extract the generated HTML content.
4. Display the generated code.
5. **NEW:** Savethe content to a file named `openai_landing_page.html`.

````python
# Let's generate HTML using Ollama API, we will use Gemma4:e4b model.

# Default message.
ollama_html_output = "<!-- Gemma 4 Generation did not completed or failed -->"
print_markdown("### Calling Gemma model using Ollama....")
try:
    response = ollama_client.chat.completions.create(
        model = 'gemma4:e4b',
        messages = [
            # No system prompt used here as instructions are in the user prompt
            {'role' : 'user', 'content' : html_prompt},
        ],
        temperature = 0.5,
    )
    ollama_html_output = response.choices[0].message.content

    # Sometimes model might wrap the code in markdown fences
    # Let's try to strip that if present
    if ollama_html_output.strip().startswith("```html"):
        lines = ollama_html_output.strip().splitlines()
        ollama_html_output = "\n".join(lines[1:-1]).strip()
    else:
        ollama_html_output = ollama_html_output.strip()

    # Display the generated HTML code
    display_html_code("Ollama (Gemma 4)", ollama_html_output)

    # Lets save the output to a file
    file_path = "ollama_landing_page.html"

    with open(file_path, 'w', encoding = 'utf-8') as f:
        f.write(ollama_html_output)
    print_markdown(f"Successfully saved Ollama/Gemma Output to `{file_path}`")


except Exception as e:
    print_markdown(f"Error calling Ollama/Gemma API : {e}")
    ollama_html_output = f"<!== Error Calling Ollama API: {e} -->"
````

### Generate HTML Landing Page using OpenAI's Model

Now, let's perform the same task using OpenAI's GPT model.
We Will: 
1. Call the `openai_client.chat.creations.create` method on our configured `GPT-4o-mini` model
2. Pass the `html_prompt`.
3. Extract the generated text (GPT's response object has a `.text` attribute).
4. Display the generated code.
5. Save the content to `openai_landing_page.html`.

````python
# Define the core prompt for the LLMs.
# We explicitly ask for HTML code and specify the file name 'index.html'
html_prompt = f"""
You are a helpful AI Assistant acting as a front-end web developer.
Your task is to generate the complete HTML code for a simple, clean, and professional-looking landing page (index.html) for a new startup.

Startup Name : {startup_name}
Concept : {startup_concept}

Please generate ONLY the full HTML code, starting with <!DOCTYPE html> and ending with </html>.
Create a modern, visually appealing landing page with the following:

-- Don't include images in the code. Raw html code with inline css for styling.

1. A sleek header with the startup name in a bold, modern font and a compelling tagline
2. A hero section with a clear value proposition and call-to-action button
3. A features section highlighting 3-4 key benefits with icons or simple visuals
4. A "How it Works" section with numbered steps
5. A testimonials section with fictional customer quotes
6. A pricing section with at least two tiers
7. A professional footer with navigation links and social media icons

Use inline CSS for styling with a modern color palette (primary, secondary, and accent colors). 
Include responsive design elements, subtle animations, and whitespace for readability.
Emphasize AI capabilities, ease of use, and business benefits throughout the copy.
Focus on conversion-optimized marketing messages that highlight pain points and solutions.

Do not include any explanations before or after the code block. Just provide the raw HTML code.
"""
print_markdown("**Core Prompt defined for the LLms:**")
print_markdown(f"> {html_prompt}")
````

````python
openai_html_output = "<-- OpenAI Generation not run or failed -->"

print_markdown("### Calling OpenAI's API")
try:
    # OpenAI API call structure
    response = openai_client.chat.completions.create(
        model = 'gpt-4o-mini',
        messages = [{
            'role' : 'user',
            'content' : html_prompt
        }],
        temperature = 0.5
    )
    openai_html_output = response.choices[0].message.content

    # Sometimes OpenAI might wrap the code in markdown fences
    # Let's try to strip that if present
    if openai_html_output.strip().startswith("```html"):
        lines = openai_html_output.strip().splitlines()
        openai_html_output = "\n".join(lines[1:-1]).strip()
    else:
        openai_html_output = openai_html_output.strip()

    # Display the generated HTML code
    display_html_code("OpenAI (gpt-4o-mini)", openai_html_output)

    # Let's Save the output to a file
    file_path = "openai_landing_page.html"

    with open(file_path, "w", encoding="utf-8") as f:
        f.write(openai_html_output)
    print_markdown(f"Successfully saved OpenAI output to `{file_path}`")


except Exception as e:
    print_markdown(f"Error calling OpenAI API: {e}")
    openai_html_output = f"<!-- Error calling OpenAI API: {e} -->"
````
