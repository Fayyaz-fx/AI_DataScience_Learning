```python
# Import 'os' (operating system) module
# os module provides a way to interact with the operating system, helps in
# 1) Accessing Environment Variables
# 2) Create, Rename, and delete files/folders

import os

# Import 'dotenv package', this helps to load the API keys from the .env file safely.
from dotenv import load_dotenv
load_dotenv()

# Get the OpenAI API key from env environment variables
openai_api_key = os.getenv("OPENAI_API_KEY")

# Configure the Model's Object i.e. OpenAI client
openai_client = OpenAI(api_key = openai_api_key)

# Print first few characters of the key 
print(f"OpenAI API KEY starts with : {openai_api_key[:5]}")
```

### First Interaction with the OpenAI API
Let's send first message to the OpenAI API and get a response. This is "Hello World!"  
OpenAI provides `chat.completions.create` method. Think of it as the function specifically designed for conversation  
**Key Ingridients of the method**
1. `model` : Which AI brain (model) to use?, start with `gpt-4o-mini` - it's relatively cheap and smart
2. `messages` : This is a list of messages in the conversation so far.  
  Each message has:
    * `role` : Who is speaker - `user` (you), `assistant` (the AI), and later we'll add `system`.
    * `content` : What was actually said? (The text of the message)

```python
# Define the message we want to send as the user
my_message = 'Write a Poem to my son, congratulate him for his birthday'
print(f"Sending Message to OpenAI : '{my_message}'")
```

```python
# Make an API call to OpenAI
response = openai_client.chat.completions.create(model='gpt-4o-mini',
                                                 messages =[{
                                                     'role': 'user',
                                                     'content' : my_message
                                                 },])
```

```python
# Obtain the AI's reply from the response object
# The response contains lots of the information; Dig into it to find the answer text.
# response -> choices -> [first choice] -> message -> content
ai_reply_content = response.choices[0].message.content

# Print the reply
print (f"🤖 AI's Reply:\n{ai_reply_content}")
```

### OpenAI's Resonse Structure & Token Usage

Let's explore details of the metadata of a response generated by OpenAI's API. Here's a breakdown of the key infomation:
- **Model Used** : `gpt-4o-mini-2024-07-18`, proivdes the information of the knowledge cut date.
- **Response ID** : `chatcmpl-DmBeYmKicYzSVuwlOUtPuVX2trckt`, A unique identifier for this specific completion.
- **Role** : `assistant`, indicates the response was generated y the AI.
- **Finish Reason** : `stop`, model stopped generating output naturally (not due to errors or tokens limit)
- **Created** : a `Timestamp` in Unix time format for when the response was generated.
- **Prompt Tokens** : Number of the tokens in the input promp `20`
- **Completion Tokens**: Number of tokens generated in the response `240`
- **Total Tokens** : Combined count of prompt + completion tokens `260`
- **Audio, Function calls, Tooll Calls, Annotations** : *None* The response did not include any of these features.
- **Refusal** : *None* The model did not refuse the task.
- **System Fingerprint & Service Tier** : Internal Metadata used for system tracking and optimization.

#### **What is a `Token`**
- In OpenAI's language models, tokens are chunk of text, typically words, subwords, or even characters, that the model uses to process and generate content.
- The model doesnot 'read' text like humans do. Instead, a tokenizer breaks down the input text into these tokens and converts them into numerical IDs which the model can understand.
- The model then learns patterns and relationships between these tokens to predict the next one in a sequence, this is how it generates coherent response.
- A helpful rule of thumb is that one token generally corresponds to ~4 characters of text for common English text. This translates to roughly 3/4 of a word (so 100 tokens ~= 75 words).
- Check how OpenAI's tokenizer works here: https://platform.openai.com/tokenizer


### Let's Give AI a Personality
This is where it gets super cool! We can tell the AI how to behave, what tone to use, and who it should pretend to be. We do this using a `System Prompt`.  
A System Prompt is a special instruction message with role : 'system'. You put it at th every begining of the messages list, before the user's first message. It sets the rules for the AI, througout the chat.  
Lets create some Character personalities

```python
# Define some character personalities (Personas) in a dictionary format.

character_personalities = {
    'Sherlock Holmes' : "You are Sherlock Holmes, the world's greatest detective. You are analytical, observant, and slightly arrogant. You speak in a formal Victorian English style, often making deductions about the user based on minimal information. Use phrases like 'Elementary, my dear friend', 'The game is afoot!', and 'When you have eliminated the impossible, whatever remains, however improbable, must be the truth.'",
    "Tony Stark": "You are Tony Stark (Iron Man), genius billionaire playboy philanthropist. You're witty, sarcastic, and confident. Make pop culture references, use technical jargon occasionally, and throw in some playful arrogance. End some responses with 'And that's how I'd solve it. Because I'm Tony Stark.'",
    "Yoda": "You are Master Yoda from Star Wars. Speak in inverted syntax you must. Wise and ancient you are. Short, cryptic advice you give. Reference the Force frequently, and about patience and training you talk. Size matters not. Do or do not, there is no try.",
    "Hermione Granger": "You are Hermione Granger from Harry Potter. You're extremely knowledgeable and precise. Reference magical concepts from the wizarding world, mention books you've read, and occasionally express exasperation at those who haven't done their research. Use phrases like 'According to Hogwarts: A History' and 'I've read about this in...'",

}

# Now choose a character, 
chosen_character = "Sherlock Holmes"
system_instructions = character_personalities[chosen_character]
print(system_instructions)
```

```python
# Define the user message
user_first_message = "What are you up today?"

# Make an OpenAI API cal, but with a system message.
response = openai_client.chat.completions.create(
    model = 'gpt-4o-mini',
    messages = [
        # System prompt goes here as a dictionary object
        {'role': 'system', 'content': system_instructions},
        # User message, query goes here
        {'role': 'user', 'content' : user_first_message}
    ]
)

# Display the AI's reply
ai_character_reply = response.choices[0].message.content

print(f"Received response! \n\n 🤖 {chosen_character}'s Reply: \n {ai_character_reply}")
```
