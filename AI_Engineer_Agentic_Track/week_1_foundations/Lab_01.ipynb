```python
# And now - all the mportant import statement
# Even for other LLM providers like Gemini, deepseek, still use OpenAI import 
# OpenAI is python client library developed by OpenAI and it's a wrapper library only.
from openai import OpenAI
```

```python
# And now we'll create an instance of the OpenAI class
# If you are not useing OpenAI, you just need slight modification, base url and api key inside the class initializer
openai = OpenAI()
```

```python
# create a list of messages 
messages = [
    {
        'role' : 'user',
        'content': "What is 2 + 2 ?"
    }
    ]
```

```python
# And now call it!
# proivde the model name and list of messages to the OpenAI's Python client library instnce

response = openai.chat.completions.create(
    model = 'gpt-4o-mini',
    messages= messages
)

print(response.choices[0].message.content)
```

```python
# And now - let's ask for a bit challening question.
question = "plese propse a hard, challenging question to assess someone's IQ. Respond with the question only."
messages = [
    {
        'role' : 'user',
        'content' : question
    }
]
```

```python
# Ask it again
response = openai.chat.completions.create(
    model = 'gpt-4o-mini',
    messages = messages
)

answer = response.choices[0].message.content
```

```python
business = ""
pain_point = ""
```
